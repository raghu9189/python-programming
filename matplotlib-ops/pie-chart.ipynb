```python
import numpy as np
from matplotlib import pyplot as plt
```

```python
# Categories and their corresponding values (excluding Total)
exp_labels = ['Home Rent', 'Food', 'Phone/Internet Bill', 'Car', 'Other Utilities']
exp_values = [1400, 600, 300, 410, 250]
```
