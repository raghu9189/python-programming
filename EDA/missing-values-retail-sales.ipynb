```python
import pandas as pd
import numpy as np
import seaborn as sns
from matplotlib import pyplot as plt
```

```python
df = pd.read_csv('retail_store_sales.csv')
df.head()
```

```python
df.info()
```

```python
df.shape
```

```python
df.isnull().sum()
```

```python
mode_value = df['Item'].mode()[0]
mode_value
```

```python
df['Item'] = df['Item'].fillna(mode_value)
```

```python
df.isnull().sum()
```

```python
df[df['Item'] == 'Item_2_BEV']
```

```python
price_mean_value = float(df['Price Per Unit'].mean())
```

```python
df['Price Per Unit'] = df['Price Per Unit'].fillna(price_mean_value)
```

```python
plt.subplot(1,1,1)
sns.boxenplot(df['Price Per Unit'])
```

```python
# no outliers out there
```

```python
quantity_mean_value = int(df['Quantity'].mean())
df['Quantity'] = df['Quantity'].fillna(quantity_mean_value)
quantity_mean_value
```

```python
# Total Spent
total_spent_mean_value = float(df['Total Spent'].mean())
df['Total Spent'] = df['Total Spent'].fillna(total_spent_mean_value)
total_spent_mean_value
```

```python
discount_mode_value = df['Discount Applied'].mode()[0]
```

```python
df['Discount Applied'] = df['Discount Applied'].fillna(discount_mode_value)
```
