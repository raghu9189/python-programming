```python
import re
# Example: Assuming your column is df['age']
def parse_age_to_years(val):
    if pd.isna(val):
        return None
    
    val = str(val).strip().upper()
    
    # Match the numeric portion and optional unit (e.g., '15 M', '21 D', '48')
    match = re.match(r"^(\d+(?:\.\d+)?)\s*([A-Z])?$", val)
    if not match:
        return None  # or handle unexpected formatting
    
    num, unit = match.groups()
    num = float(num)
    
    if unit == 'M':
        return round(num / 12, 2)
    elif unit == 'D':
        return round(num / 365, 2)
    else:
        return num  # Plain years

# Apply to your series
df['cleaned_age'] = df['age'].apply(parse_age_to_years)
```
