```python
# importing necessary packages
import pandas as pd
import numpy as np
import seaborn as sns
from matplotlib import pyplot as plt 
```

```python
# we are gonna work with tips dataset 
data = sns.load_dataset('tips')
data.head()
```

```python
# Compare weather Male are paying the highest bill Or female on average.
plt.figure(figsize=(6,4))
ax = sns.barplot(x=data['sex'], y=data['total_bill'])
ax.bar_label(ax.containers[0])
plt.show()
```
