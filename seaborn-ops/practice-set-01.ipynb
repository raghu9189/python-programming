```python
# importing necessary packages
import pandas as pd
import numpy as np
import seaborn as sns
from matplotlib import pyplot as plt 
```

```python
# we are gonna work with tips dataset 
data = sns.load_dataset('tips')
data.head()
```

```python
# Compare weather Male are paying the highest bill Or female on average.
plt.figure(figsize=(6,4))
ax = sns.barplot(x=data['sex'], y=data['total_bill'])
ax.bar_label(ax.containers[0])
plt.show()
```

```python
# We want to get the sum of total bill By both the genders.
agg_data = data.groupby('sex')['total_bill'].agg('sum').reset_index()
agg_data
```

```python
ax = sns.barplot(x="sex", y="total_bill", data=agg_data)
ax.bar_label(ax.containers[0])
plt.show()
```

```python
# What is the average total bill by male smoker? and non-male smoker and female smoker? and non-female smoker
ax = sns.barplot(x=data["sex"], y=data["total_bill"], data=agg_data, hue=data["smoker"])
ax.bar_label(ax.containers[0])
ax.bar_label(ax.containers[1])
plt.show()
```

```python
# count smoker and non-smoker
ax = sns.countplot(x=data['smoker'], hue=data['sex'])
ax.bar_label(ax.containers[0])
ax.bar_label(ax.containers[1])
plt.show()
```
